# Integración Numérica: Regla del Trapecio y Simpson

## Objetivo

Dada una integral definida:

$$
I(f)=\int_a^b f(x)dx
$$

el objetivo es aproximar numéricamente el valor del área bajo la curva cuando la integral exacta es difícil o imposible de calcular.

---

# Idea general: Newton-Cotes Cerradas

Las reglas de Newton-Cotes aproximan la función mediante polinomios construidos sobre nodos igualmente espaciados.

Los nodos se definen como:

$$
x_k=a+kh
$$

donde:

$$
h=\frac{b-a}{n}
$$

es el tamaño de paso.

---

# Regla del Trapecio

La idea es aproximar la curva mediante segmentos rectos.

Geométricamente:

- Se unen dos puntos consecutivos mediante una línea recta
- El área se aproxima usando trapecios
- Mayor cantidad de subintervalos generalmente mejora la aproximación

## Fórmula compuesta del trapecio

$$
I(f)\approx
h
\left[
\frac{f(a)+f(b)}{2}
+
\sum_{k=1}^{n-1}f(x_k)
\right]
$$

donde:

- $h$ = tamaño de paso
- $n$ = número de subintervalos

## Error del Trapecio

El error tiene orden:

$$
E_T=O(h^2)
$$

Por lo tanto:

**Disminuir $h$ mejora la aproximación.**

---

# Regla de Simpson

La idea es aproximar la función usando parábolas en lugar de rectas.

Esto normalmente produce una aproximación más precisa.

## Restricción importante

Simpson requiere:

$$
n \text{ par}
$$

porque trabaja agrupando intervalos de dos en dos.

## Fórmula compuesta de Simpson

$$
I(f)
\approx
\frac{h}{3}
\left[
f(x_0)
+
4\sum f(x_{impares})
+
2\sum f(x_{pares})
+
f(x_n)
\right]
$$

donde:

- Los nodos impares tienen peso 4
- Los nodos pares tienen peso 2

## Error de Simpson

El error tiene orden:

$$
E_S=O(h^4)
$$

Por lo tanto:

**Simpson generalmente converge más rápido que Trapecio.**

---

# Comparación rápida

| Método | Aproximación | Orden Error | Restricción |
|---|---:|---:|---:|
| Trapecio | Rectas | $O(h^2)$ | Ninguna |
| Simpson | Parábolas | $O(h^4)$ | $n$ par |

---
Mientras mayor sea el orden, más rápido disminuye el error al refinar la malla.
