## 3. Definición del problema

Se define el lado derecho de la ecuación diferencial

$$
f(t,T)
=
-k(T-T_a),
$$

así como la solución exacta

$$
T(t)
=
T_a
+
(T_0-T_a)e^{-kt},
$$

la cual servirá como referencia para calcular los errores absoluto y relativo de todos los métodos numéricos implementados.

```python
# ============================================================
# 3. DEFINICIÓN DEL PROBLEMA
# ============================================================
#
# Función del lado derecho
#
#      T' = f(t,T)
#
# ============================================================

def f(t, T):
    """
    Lado derecho de la Ley de Enfriamiento de Newton.

    Parámetros
    ----------
    t : float
        Tiempo.

    T : float
        Temperatura.

    Retorna
    -------
    float
        Valor de la derivada dT/dt.
    """
    return -k*(T-Ta)

# ============================================================
# SOLUCIÓN EXACTA
# ============================================================
# Utilice esta función para verificar que la solución obtenida
# en la Parte I del laboratorio es correcta. Esta función será
# empleada posteriormente para calcular los errores numéricos.
# ============================================================

def exacta(t):
    """
    Solución exacta del problema.

    T(t)=Ta+(T0-Ta)e^{-kt}
    """
    return Ta + (T0-Ta)*np.exp(-k*t)


print("="*72)
print("PROBLEMA DEFINIDO CORRECTAMENTE")
print("="*72)
```

# 12. Cálculo de errores

Se comparan las aproximaciones numéricas con la solución exacta del problema.

Para cada método se calcularán:

- Error absoluto

$$
E_a=|T(t)-T_h(t)|
$$

- Error relativo

$$
E_r=\frac{|T(t)-T_h(t)|}{|T(t)|}.
$$

Estos indicadores permitirán evaluar la precisión de cada método numérico.

```python
# ============================================================
# 12. CÁLCULO DE ERRORES
# ============================================================

# ------------------------------------------------------------
# Error absoluto
# ------------------------------------------------------------

error_euler = np.abs(T_exacta - T_euler)
error_trap  = np.abs(T_exacta - T_trap)
error_heun  = np.abs(T_exacta - T_heun)

error_ab2   = np.abs(T_exacta - T_ab2)
error_am2   = np.abs(T_exacta - T_am2)
error_pc    = np.abs(T_exacta - T_pc)

# ------------------------------------------------------------
# Error relativo
# ------------------------------------------------------------

error_rel_euler = error_euler / np.abs(T_exacta)
error_rel_trap  = error_trap  / np.abs(T_exacta)
error_rel_heun  = error_heun  / np.abs(T_exacta)

error_rel_ab2 = error_ab2 / np.abs(T_exacta)
error_rel_am2 = error_am2 / np.abs(T_exacta)
error_rel_pc  = error_pc  / np.abs(T_exacta)

# ------------------------------------------------------------
# Evitar ceros en las gráficas semilogarítmicas
# ------------------------------------------------------------

error_rel_euler_plot = np.where(error_rel_euler > 0,
                                error_rel_euler,
                                np.nan)

error_rel_trap_plot = np.where(error_rel_trap > 0,
                               error_rel_trap,
                               np.nan)

error_rel_heun_plot = np.where(error_rel_heun > 0,
                               error_rel_heun,
                               np.nan)

error_rel_ab2_plot = np.where(error_rel_ab2 > 0,
                              error_rel_ab2,
                              np.nan)

error_rel_am2_plot = np.where(error_rel_am2 > 0,
                              error_rel_am2,
                              np.nan)

error_rel_pc_plot = np.where(error_rel_pc > 0,
                             error_rel_pc,
                             np.nan)

print("="*72)
print("ERRORES CALCULADOS CORRECTAMENTE")
print("="*72)

display(pd.DataFrame({

    "Tiempo":t,

    "Exacta":T_exacta,

    "Euler":error_euler,

    "Trapecio":error_trap,

    "Heun":error_heun

}).head())
```
