# Resumen teórico para resolver el Ejercicio 1

## 1. ¿Qué es una cuadratura?

La integración numérica busca aproximar una integral mediante una suma de evaluaciones de la función.

En lugar de calcular

$$
I(f)=\int_{-1}^{1} f(x)\,dx,
$$

se utiliza una aproximación de la forma

$$
I_n(f)=\sum_{i=0}^{n} A_i f(x_i),
$$

donde:

- $$x_i$$ son los nodos de integración.
- $$A_i$$ son los pesos.

---

## 2. Idea principal de la cuadratura gaussiana

A diferencia de las fórmulas de Newton-Cotes (Trapecio, Simpson, etc.), los nodos no están fijados previamente.

Los pesos y los nodos se eligen de manera óptima para obtener la mayor precisión posible.

---

## 3. Exactitud algebraica

Una fórmula de cuadratura tiene grado de precisión $$m$$ si integra exactamente todos los polinomios de grado menor o igual que $$m$$.

Por ejemplo, si queremos una fórmula exacta para:

$$
1,\quad x,\quad x^2,\quad x^3,
$$

debemos imponer que la cuadratura produzca exactamente el mismo valor que la integral para cada uno de estos polinomios.

---

## 4. Construcción del sistema de ecuaciones

La fórmula buscada es

$$
I_2(f)=A_0f(x_0)+A_1f(x_1).
$$

Como hay cuatro incógnitas

$$
A_0,\;A_1,\;x_0,\;x_1,
$$

debemos generar cuatro ecuaciones.

Para ello imponemos exactitud para los monomios:

### Para $$1$$

$$
\int_{-1}^{1}1\,dx=2
$$

y

$$
A_0+A_1=2.
$$

---

### Para $$x$$

$$
\int_{-1}^{1}x\,dx=0
$$

y

$$
A_0x_0+A_1x_1=0.
$$

---

### Para $$x^2$$

$$
\int_{-1}^{1}x^2\,dx
=
\frac{2}{3}
$$

y

$$
A_0x_0^2+A_1x_1^2
=
\frac{2}{3}.
$$

---

### Para $$x^3$$

$$
\int_{-1}^{1}x^3\,dx=0
$$

y

$$
A_0x_0^3+A_1x_1^3=0.
$$

---

## 5. Solución del sistema

Al resolver el sistema se obtiene:

$$
A_0=A_1=1,
$$

$$
x_0=-\frac{1}{\sqrt{3}},
\qquad
x_1=\frac{1}{\sqrt{3}}.
$$

Por tanto, la fórmula final es

$$
\int_{-1}^{1}f(x)\,dx
\approx
f\left(-\frac{1}{\sqrt{3}}\right)
+
f\left(\frac{1}{\sqrt{3}}\right).
$$

---

## 6. Teorema fundamental de la cuadratura gaussiana

Si una cuadratura utiliza

$$
n+1
$$

nodos, entonces puede alcanzar un grado de precisión máximo igual a

$$
2n+1.
$$

En este ejercicio:

$$
n+1=2,
$$

por lo que

$$
n=1.
$$

Así, el grado de precisión es

$$
2(1)+1=3.
$$

Esto significa que la fórmula integra exactamente cualquier polinomio de grado menor o igual que tres.

---



## Respuestas

### 1. Sistema de ecuaciones

Se busca una fórmula de la forma

$$
I_2(f)=A_0f(x_0)+A_1f(x_1)
$$

para aproximar

$$
I(f)=\int_{-1}^{1}f(x)\,dx.
$$

Imponiendo exactitud para

$$
1,\quad x,\quad x^2,\quad x^3,
$$

se obtiene el sistema:

$$
A_0+A_1=2,
$$

$$
A_0x_0+A_1x_1=0,
$$

$$
A_0x_0^2+A_1x_1^2=\frac{2}{3},
$$

$$
A_0x_0^3+A_1x_1^3=0.
$$

---

### 2. Solución del sistema

La solución es

$$
A_0=1,
\qquad
A_1=1,
$$

$$
x_0=-\frac{1}{\sqrt{3}},
\qquad
x_1=\frac{1}{\sqrt{3}}.
$$

Por tanto, la fórmula de cuadratura es

$$
I_2(f)=f\left(-\frac{1}{\sqrt{3}}\right)
+
f\left(\frac{1}{\sqrt{3}}\right).
$$

---

### 3. Verificación para un polinomio cúbico general

Sea

$$
p(x)=a_0+a_1x+a_2x^2+a_3x^3.
$$

La integral exacta es

$$
\int_{-1}^{1}p(x)\,dx
=
\int_{-1}^{1}
(a_0+a_1x+a_2x^2+a_3x^3)\,dx.
$$

Como

$$
\int_{-1}^{1}1\,dx=2,
$$

$$
\int_{-1}^{1}x\,dx=0,
$$

$$
\int_{-1}^{1}x^2\,dx=\frac{2}{3},
$$

$$
\int_{-1}^{1}x^3\,dx=0,
$$

se tiene

$$
\int_{-1}^{1}p(x)\,dx
=
2a_0+\frac{2}{3}a_2.
$$

Ahora, usando la cuadratura:

$$
I_2(p)=
p\left(-\frac{1}{\sqrt{3}}\right)
+
p\left(\frac{1}{\sqrt{3}}\right).
$$

Sustituyendo,

$$
I_2(p)=
2a_0+\frac{2}{3}a_2.
$$

Por tanto,

$$
I_2(p)=\int_{-1}^{1}p(x)\,dx.
$$

Entonces la fórmula es exacta para todo polinomio de grado menor o igual que tres.

---

### 4. Aplicación a la integral de $$e^x$$

Se desea aproximar

$$
I=\int_{-1}^{1}e^x\,dx.
$$

Aplicando la cuadratura:

$$
I_2=
e^{-\frac{1}{\sqrt{3}}}
+
e^{\frac{1}{\sqrt{3}}}.
$$

El valor exacto es

$$
I=e^1-e^{-1}.
$$

Numéricamente,

$$
I \approx 2.3504023873,
$$

mientras que

$$
I_2 \approx 2.3426960879.
$$

---

### 5. Error absoluto y error relativo

El error absoluto es

$$
E_{\text{abs}}=|I-I_2|.
$$

Sustituyendo,

$$
E_{\text{abs}}
=
|2.3504023873-2.3426960879|.
$$

Por tanto,

$$
E_{\text{abs}}\approx 0.0077062994.
$$

El error relativo es

$$
E_{\text{rel}}=
\frac{|I-I_2|}{|I|}.
$$

Entonces,

$$
E_{\text{rel}}
\approx
\frac{0.0077062994}{2.3504023873}.
$$

Por tanto,

$$
E_{\text{rel}}\approx 0.0032787142.
$$

---

### 6. Grado de precisión algebraica

La fórmula tiene dos nodos:

$$
x_0,\quad x_1.
$$

En cuadratura gaussiana, una fórmula con

$$
n+1
$$

nodos alcanza grado de precisión máximo

$$
2n+1.
$$

En este caso,

$$
n+1=2,
$$

por tanto,

$$
n=1.
$$

Entonces el grado de precisión es

$$
2n+1=2(1)+1=3.
$$

Por eso la fórmula integra exactamente todos los polinomios de grado menor o igual que tres.

````python
import sympy as sp                    # Biblioteca para cálculo simbólico
import numpy as np                    # Biblioteca para cálculo numérico

A0, A1, x0, x1 = sp.symbols('A0 A1 x0 x1')  # Define las incógnitas simbólicas

eq1 = sp.Eq(A0 + A1, 2)              # Exactitud para p(x)=1
eq2 = sp.Eq(A0*x0 + A1*x1, 0)        # Exactitud para p(x)=x
eq3 = sp.Eq(A0*x0**2 + A1*x1**2, 2/3) # Exactitud para p(x)=x²
eq4 = sp.Eq(A0*x0**3 + A1*x1**3, 0)  # Exactitud para p(x)=x³

sol = sp.solve([eq1,eq2,eq3,eq4],    # Resuelve el sistema de ecuaciones
               [A0,A1,x0,x1],        # Variables desconocidas
               dict=True)            # Devuelve la solución como diccionario

print(sol)                           # Muestra todas las soluciones encontradas

# seleccionar una solución
s = sol[0]                           # Toma la primera solución de la lista

A0 = float(s[A0])                    # Convierte A0 a número decimal
A1 = float(s[A1])                    # Convierte A1 a número decimal
x0 = float(s[x0])                    # Convierte x0 a número decimal
x1 = float(s[x1])                    # Convierte x1 a número decimal

# integral exacta
I_exacta = np.exp(1)-np.exp(-1)      # Calcula ∫_{-1}^{1} e^x dx = e - e^{-1}

# cuadratura gaussiana
I_gauss = A0*np.exp(x0)+A1*np.exp(x1) # Aplica la fórmula de cuadratura obtenida

error_abs = abs(I_exacta-I_gauss)    # Calcula el error absoluto
error_rel = error_abs/abs(I_exacta)  # Calcula el error relativo

print("Integral exacta =",I_exacta)  # Muestra el valor exacto
print("Cuadratura =",I_gauss)        # Muestra la aproximación gaussiana
print("Error absoluto =",error_abs)  # Muestra el error absoluto
print("Error relativo =",error_rel)  # Muestra el error relativo
````

## 2. Cuadratura de Gauss-Legendre

La fórmula general es

$$
I_n(f)
=
\sum_{i=1}^{n}
A_i f(x_i),
$$

donde:

- $$x_i$$ son las raíces del polinomio de Legendre $$P_n(x)$$.
- $$A_i$$ son los pesos asociados.

La aproximación de la integral es

$$
\int_{-1}^{1}f(x)\,dx
\approx
\sum_{i=1}^{n}

$$

---

## 3. ¿Qué son los polinomios de Legendre?

Son una familia de polinomios ortogonales en el intervalo

$$
[-1,1].
$$

Los primeros son:

$$
P_0(x)=1,
$$

$$
P_1(x)=x,
$$

$$
P_2(x)=\frac12(3x^2-1),
$$

$$
P_3(x)=\frac12(5x^3-3x).
$$

Los nodos de Gauss-Legendre se obtienen calculando las raíces de estos polinomios.

---

## 4. Grado de precisión algebraica

La principal ventaja de Gauss-Legendre es que con

$$
n
$$

nodos puede integrar exactamente todos los polinomios de grado menor o igual que

$$
2n-1.
$$

Por ejemplo:

| Nodos | Grado de precisión |
|---------|---------|
| 1 | 1 |
| 2 | 3 |
| 3 | 5 |
| 4 | 7 |
| n | 2n−1 |

Esto explica por qué suele ser más precisa que otras reglas con el mismo número de evaluaciones.

---

## 5. Error de cuadratura

El error se define como

$$
E_n
=
I-I_n,
$$

donde:

- $$I$$ es el valor exacto (o una aproximación muy precisa).
- $$I_n$$ es la aproximación obtenida con $$n$$ nodos.

En la práctica se calcula

$$
|E_n|
=
|I-I_n|.
$$

Este valor permite medir la calidad de la aproximación.

---

## 6. Teorema del error de Gauss-Legendre

Si

$$
f\in C^{2n}[-1,1],
$$

entonces el error tiene la forma

$$
E_n(f)
=
\frac{
2^{2n+1}(n!)^4
}{
(2n+1)\left[(2n)!\right]^3
}
f^{(2n)}(\eta),
$$

para algún

$$
\eta\in(-1,1).
$$

La idea importante es que:

- Si la función es suave, el error disminuye muy rápido.
- Al aumentar el número de nodos, la precisión mejora significativamente.

# Resumen teórico para resolver el Ejercicio 3

## 1. Objetivo del ejercicio

El objetivo es comparar la precisión de dos métodos de integración numérica:

- Cuadratura de Gauss-Legendre.
- Regla del Trapecio.

Ambos métodos se aplican a la integral

$$
I=\int_{-1}^{1} e^{-x^2}\,dx.
$$

Se analizará cuál produce un error menor utilizando el mismo número de evaluaciones de la función.

---

## 2. La integral exacta

La integral a aproximar es

$$
I=\int_{-1}^{1} e^{-x^2}\,dx.
$$

La función

$$
f(x)=e^{-x^2}
$$

es suave, continua y derivable infinitas veces.

Este tipo de funciones suele ser favorable para las cuadraturas gaussianas.

---

## 3. Cuadratura de Gauss-Legendre

La fórmula general es

$$
\int_{-1}^{1}f(x)\,dx
\approx
\sum_{i=1}^{n}
A_i f(x_i),
$$

donde:

- $$x_i$$ son las raíces del polinomio de Legendre $$P_n(x)$$.
- $$A_i$$ son los pesos asociados.

---

## 4. Fórmula de tres nodos

Para tres nodos:

$$
x_1=-\sqrt{\frac35},
\qquad
x_2=0,
\qquad
x_3=\sqrt{\frac35}.
$$

Los pesos son

$$
A_1=\frac59,
\qquad
A_2=\frac89,
\qquad
A_3=\frac59.
$$

La fórmula queda

$$
I_G=
\frac59f\left(-\sqrt{\frac35}\right)
+
\frac89f(0)
+
\frac59f\left(\sqrt{\frac35}\right).
$$

---

## 5. ¿Por qué se utilizan esos nodos?

Los nodos son las raíces del polinomio de Legendre

$$
P_3(x).
$$

Estos puntos se eligen para maximizar la precisión de la cuadratura.

No son equiespaciados.

---

## 6. Grado de precisión algebraica

Una cuadratura de Gauss-Legendre con

$$
n
$$

nodos integra exactamente todos los polinomios de grado menor o igual que

$$
2n-1.
$$

Para tres nodos:

$$
2(3)-1=5.
$$

Por tanto integra exactamente cualquier polinomio de grado

$$
\le 5.
$$

---

## 7. Regla del Trapecio

La regla compuesta del trapecio con tres nodos utiliza:

$$
x_0=-1,
\qquad
x_1=0,
\qquad
x_2=1.
$$

Estos puntos sí son equiespaciados.

La fórmula es

$$
I_T
=
\frac{h}{2}
\Big(
f(x_0)+2f(x_1)+f(x_2)
\Big),
$$

donde

$$
h=1.
$$

Por tanto

$$
I_T
=
\frac12
\Big(
f(-1)+2f(0)+f(1)
\Big).
$$

---

## 8. Diferencia principal entre ambos métodos

### Trapecio

- Utiliza nodos equiespaciados.
- Es fácil de construir.
- Tiene menor precisión.

### Gauss-Legendre

- Utiliza nodos óptimos.
- Maximiza la precisión.
- Obtiene errores mucho menores.

---

## 9. Error absoluto

Una vez obtenidas las aproximaciones se calcula

$$
E=
|I-I_{aprox}|.
$$

Si se usa Gauss-Legendre:

$$
E_G=
|I-I_G|.
$$

Si se usa Trapecio:

$$
E_T=
|I-I_T|.
$$

El método con menor error será el más preciso.

---

## 10. Comparación esperada

Debido a que Gauss-Legendre de tres nodos tiene grado de precisión cinco, se espera que:

$$
E_G<E_T.
$$

Por tanto, la cuadratura gaussiana debería proporcionar una aproximación más cercana al valor exacto.

## Fundamento teórico

La cuadratura de **Gauss-Laguerre** es un método de integración numérica diseñado para aproximar integrales de la forma

$$
\int_{0}^{\infty} e^{-x} f(x),dx.
$$

Este método utiliza nodos y pesos especialmente seleccionados a partir de los polinomios de Laguerre, lo que permite obtener aproximaciones muy precisas con un número reducido de puntos de evaluación.

La fórmula general de la cuadratura de Gauss-Laguerre es

$$
\int_{0}^{\infty} e^{-x} f(x),dx
\approx
\sum_{i=1}^{n} w_i f(x_i),
$$

donde:

* $x_i$ son los nodos (raíces del polinomio de Laguerre de grado (n)),
* $w_i$ son los pesos asociados,
* $n$ es el número de nodos utilizados.

Este método es especialmente adecuado cuando la integral contiene explícitamente el factor de peso \(e^{-x}\), ya que dicho término queda incorporado en la fórmula de cuadratura.




## Introducción teórica

La cuadratura de **Gauss-Hermite** es un método de integración numérica diseñado para aproximar integrales de la forma

$$
\int_{-\infty}^{\infty} e^{-x^2} f(x),dx.
$$

Este método pertenece a la familia de las cuadraturas gaussianas y utiliza nodos y pesos especialmente seleccionados para obtener la máxima precisión posible con un número reducido de puntos de evaluación.

Los nodos de la cuadratura corresponden a las raíces de los polinomios de Hermite, mientras que los pesos se calculan de manera que la fórmula sea exacta para polinomios de grado lo más alto posible. Con (n) nodos, la cuadratura de Gauss-Hermite integra exactamente todos los polinomios de grado menor o igual que (2n-1).

La aproximación general viene dada por

$$
\int_{-\infty}^{\infty} e^{-x^2} f(x)dx
\approx
\sum_{i=1}^{n} w_i,f(x_i),
$$

donde:

* $x_i$ son los nodos (raíces de los polinomios de Hermite),
* $w_i$ son los pesos asociados,
* $n$ es el número de nodos utilizados.

Este método es especialmente útil cuando el integrando contiene el factor de peso \(e^{-x^2}\), ya que dicho término queda incorporado en la fórmula de integración.
